## 3. Success metric

*One metric you can defend. What number means 'good'?*

I would use F1-score as the main success metric because this is a classification problem with a declining and non-declining outcome.

F1-score considers both precision and recall. Precision tells us how often the content predicted as declining is actually declining, while recall tells us how much of the actually declining content we are able to identify.

A higher F1-score would mean the model is better at identifying declining content while balancing false alarms and missed declining content.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule could identify declining content using one simple condition, such as flagging content when a performance measure falls below a chosen threshold. However, content performance can depend on many factors at the same time.

ML could learn patterns across multiple features, such as content type, search volume, word count, content age, freshness, and historical performance. This makes ML more flexible than choosing one fixed threshold.

The output would support a real content action: helping an editor identify content that may need review or updating. The editor would still make the final decision rather than automatically changing the content.
